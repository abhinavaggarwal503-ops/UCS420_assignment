```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save DataFrame as CSV file
df.to_csv("employees.csv", index=False)

# Display the dataset
print(df)
```

```python
# a) Shape of DataFrame
print("a) Shape of DataFrame:")
print(df.shape)

# b) Summary of DataFrame
print("\nb) DataFrame Information:")
df.info()

# c) Generate descriptive statistics
print("\nc) Descriptive Statistics:")
print(df.describe())

# d) Display first 5 rows and last 3 rows
print("\nd) First 5 rows:")
print(df.head(5))

print("\nLast 3 rows:")
print(df.tail(3))

```

```python
# e) Calculate the following statistics

# i. Average salary of employees
average_salary = df["Salary"].mean()
print("i) Average Salary:", average_salary)

# ii. Total bonus paid to all employees
total_bonus = df["Bonus"].sum()
print("ii) Total Bonus:", total_bonus)

# iii. Youngest employee's age
youngest_age = df["Age"].min()
print("iii) Youngest Employee's Age:", youngest_age)

# iv. Highest performance rating
highest_rating = df["Rating"].max()
print("iv) Highest Performance Rating:", highest_rating)
```

```python
# f) Sort DataFrame by Salary in descending order

print("\nf) DataFrame sorted by Salary in descending order:")

sorted_df = df.sort_values(by="Salary", ascending=False)

print(sorted_df)
```

```python
# g) Categorize employees based on Rating


def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"


df["Performance_Category"] = df["Rating"].apply(categorize_rating)

print("\ng) Performance Category:")
print(df)
```

```python
# h) Identify missing values


print("\nh) Missing values in each column:")

print(df.isnull().sum())
```

```python
# i) Rename Employee_ID column to ID

df.rename(columns={"Employee_ID": "ID"}, inplace=True)

print("After renaming Employee_ID to ID:")
print(df)

# j-i) Employees with more than 5 years of experience

employees_experience = df[df["Years_of_Experience"] > 5]

print("Employees with more than 5 years of experience:")
print(employees_experience)
```

```python
# j-ii) Employees belonging to IT department

employees_IT = df[df["Department"] == "IT"]

print("Employees belonging to IT department:")
print(employees_IT)
```

```python
# k) Add Tax column

df["Tax"] = df["Salary"] * 0.10

print("DataFrame after adding Tax:")
print(df)
```

```python
# l) Save modified DataFrame to a new CSV file

df.to_csv("modified_employee.csv", index=False)

print("Modified DataFrame saved successfully.")
```
